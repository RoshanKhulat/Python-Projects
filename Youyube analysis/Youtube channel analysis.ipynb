### Find the top 5 Youtube channel having the maximum Subscriber

```python
df.columns
```

```python
# arranging the data as per the subscribers from max to min
data_top_10_yt = df.groupby('Youtuber' , as_index = False)['Subscribers'].sum().sort_values(by ='Subscribers', ascending = False)
data_top_10_yt.head()
```

```python
plt.figure(figsize = (12,5))
ax = sns.barplot(x ='Youtuber', y = 'Subscribers', data = data_top_10_yt.head())
for bar in ax.containers:
    ax.bar_label(bar)

plt.xlabel('Channel')
plt.ylabel('Sucscribers count')
plt.title('Total Subscribers by channel')
plt.show()
```

### Conclusion
#### As per the analysis we swa that the above five youtube channel having the maximum number of Subscribers

### Analyze the posted videos and view on that video

```python
plt.figure(figsize = (12,5))
sns.scatterplot(x = 'Video Count' , y = 'Video Views' , data = df )

plt.xlabel('Posted videos')
plt.ylabel('Views')
plt.title('Posted videos and views on that video')
plt.show()
```

### Check which kind of videos are mostly watched?

```python
df.columns
```

```python
data_cat = df.groupby('Category' , as_index = False)['Video Views'].sum().sort_values(by= 'Video Views' , ascending = False)
```

```python
data_cat.head()
```

```python
plt.figure(figsize = (16,5))
sns.barplot(x = 'Category' ,y = 'Video Views' , data = data_cat.head(10))

plt.title('Mostly watched videos as per the category')
plt.show()
```

### Conclusion
#### As per the analysis we swa that the Most people watch the Entertainment, Music and People & Blogs Videos. So currently this category are performing very high compition and high growth as well.
